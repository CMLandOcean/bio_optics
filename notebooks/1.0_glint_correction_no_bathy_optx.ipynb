# Two-step EnMAP inversion

**Step 1 — Glint correction**: joint 3C inversion with loose priors → glint surface Rrs → glint-corrected Rrs saved to zarr.

**Step 2 — Water quality retrieval**: water-only inversion on glint-corrected Rrs with tighter concentration priors → WQ parameter maps saved to zarr.

**Bathymetry prior (optional)**: set `BATHY_VERSION` in the config cell to activate per-pixel `zB` priors from a pre-computed depth product.  When `None` the depth prior is a global scalar from `params_3C`/`params_wq`.

### Glint forward model (2D)

```python
def forward_glint_2D(fit_param_ds, parameters, wavelengths, pre_3C):
    """Compute glint surface Rrs from OE-fitted glint params and precomputed irradiances.

    Returns array of shape (n_bands, n_y, n_x) matching rrs dims.

    Accepts both old dataset format (per-variable) and new dask_oe_engine.to_dataset
    format (x_hat with a 'param' dimension).
    """
    n2 = float(parameters['n2'].value)
    if 'rho_L' in parameters:
        rho_L = float(parameters['rho_L'].value)
    else:
        rho_L = air_water.fresnel(
            parameters['theta_view'].value, n1=parameters['n1'].value, n2=n2
        )

    # 1D spectral arrays — kept as (n_bands,) to avoid large broadcasts
    Ed_d  = np.array(pre_3C['Ed_d'],  dtype=float)
    Ed_sr = np.array(pre_3C['Ed_sr'], dtype=float)
    Ed_sa = np.array(pre_3C['Ed_sa'], dtype=float)
    Ls_Ed = np.array(pre_3C['Ls_Ed'], dtype=float)
    Ed    = Ed_d + Ed_sr + Ed_sa

    def _get_param(name):
        """Extract (n_y, n_x) array from either old or new dataset format."""
        if name in fit_param_ds:
            return fit_param_ds[name].values                      # old: per-variable
        return fit_param_ds['x_hat'].sel(param=name).values       # new: x_hat[..., param]

    # Spatial arrays (n_y, n_x, 1) — the trailing dim broadcasts against (n_bands,)
    g_dd  = _get_param('g_dd')[...,  np.newaxis]
    g_dsr = _get_param('g_dsr')[..., np.newaxis]
    g_dsa = _get_param('g_dsa')[..., np.newaxis]
    d_r   = _get_param('d_r')[...,   np.newaxis]

    fd_d       = float(parameters['fd_d'].value)
    fd_s       = float(parameters['fd_s'].value)
    theta_view = float(parameters['theta_view'].value)

    # Broadcasting: (n_y, n_x, 1) × (n_bands,) → (n_y, n_x, n_bands) on-the-fly
    L_s = sky_radiance.L_s(
        fd_d, g_dd, Ed_d,
        fd_s, g_dsr, Ed_sr,
              g_dsa, Ed_sa,
    )

    R_rs_surface  = Rrs_surf(L_s, Ed, rho_L, d_r)
    R_rs_surface += air_water.fresnel(theta_view, n2=n2) * Ls_Ed

    return R_rs_surface.transpose(2, 0, 1)  # (n_bands, n_y, n_x)
```

## Step 2 — Water quality retrieval

### Parameters (water-only model, tighter priors)

Same IOP/geometry names as the 3C model — glint params removed.  
Concentration priors tightened from σ=1.0 to σ=0.7 (≈factor-of-2 uncertainty instead of factor-of-e).

```python
import configparser
from pathlib import Path

config = configparser.ConfigParser()
config.read('../../config.ini')
credentials = {key: value.strip() for key, value in config['Credentials'].items()}

store = new_data_store(
    's3',
    max_depth=5,
    root="coastal-cubes/sek/",
    storage_options=dict(
        anon=False,
        key=credentials['s3_client_id'],
        secret=credentials['s3_client_secret'],
    )
)

INPUT_PREFIX  = "helsinki/L2A_land/"
OUTPUT_PREFIX = "helsinki/temp/"

# Set to a version string (e.g. "v0.1") to activate per-pixel zB priors from
# a pre-computed bathymetry product stored at {OUTPUT_PREFIX}{scene_id}-bathymetry-{BATHY_VERSION}.zarr.
# Set to None for a global scalar zB prior (no bathymetry required).
BATHY_VERSION = None

SCENE_IDS = [
    "ENMAP01-____L2A-DT0000158841_20251019T101424Z_002_V010505_20260206T113145Z",
]
```

```python
TILE_SIZE  = 4096   # pixels per tile
MAX_STEPS  = 100    # optimistix: max GN iterations (exits early on convergence)

for scene_id in SCENE_IDS:

    rrs_path = f"{OUTPUT_PREFIX}{scene_id}-Rrs.zarr"
    wq_path  = f"{OUTPUT_PREFIX}{scene_id}-wq_params.zarr"

    existing   = set(store.get_data_ids())
    need_step1 = rrs_path not in existing
    need_step2 = wq_path  not in existing

    if not need_step1 and not need_step2:
        print(f'{scene_id}: all outputs present, skipping.')
        continue

    print(f'Processing {scene_id}.')

    img         = store.open_data(f"{INPUT_PREFIX}{scene_id}.zarr")
    scene_crs   = img.rio.crs or CRS.from_wkt(img.spatial_ref.attrs['crs_wkt'])
    wavelengths = img.wavelength.values[:80]

    # --- Optional: per-pixel depth prior from bathymetry product ---
    depth_map = None
    bathy_sea  = None
    if BATHY_VERSION is not None:
        bathy      = store.open_data(f"{OUTPUT_PREFIX}{scene_id}-bathymetry-{BATHY_VERSION}.zarr")
        depth_map  = np.maximum(-bathy['elevation'].values, 0.1)
        bathy_sea  = bathy['elevation'] < 0

    # =========================================================================
    # STEP 1 — Glint correction (oe_engine_optx: lax.while_loop, early exit)
    # =========================================================================
    if need_step1:
        print('  Step 1: glint correction ...')

        refl = img['reflectance'].isel(band=slice(0, 80))
        rrs  = (refl.where(refl > -32768) / 10_000) / np.pi

        cloud_mask = (img['cloud'] == 1) | (img['cirrus'] == 1) | (img['haze'] == 1)
        rrs = rrs.where(~cloud_mask)

        MIN_WATER_PIXELS = 100
        _wc = store.open_data(f"{OUTPUT_PREFIX}{scene_id}-worldcover.zarr")
        _wf = _wc['water_fraction'].values
        _labeled, _ = ndi.label(_wf >= 0.5)
        _sizes = np.bincount(_labeled.ravel())
        _sizes[0] = 0
        wc_ocean_mask = xr.DataArray(
            np.isin(_labeled, np.where(_sizes >= MIN_WATER_PIXELS)[0]),
            coords=_wc['water_fraction'].coords,
            dims=_wc['water_fraction'].dims,
        )
        ocean_mask = wc_ocean_mask & bathy_sea if bathy_sea is not None else wc_ocean_mask
        rrs = rrs.where(ocean_mask)

        pre_3C = albert_mobley_3C_jax.precompute(
            wavelengths, fresh=False,
            theta_sun=float(params_3C['theta_sun'].value),
            P=1013.25, AM=1, RH=60, H_oz=0.38, WV=2.5, alpha=1.317, beta=0.2606,
        )
        f_vec_3C = albert_mobley_3C_jax.make_forward_vec(list(params_3C.keys()), pre_3C)
        setup_3C = oe_engine.build_inversion(
            params_3C, f_vec_3C, sigma_a_3C, log_params=log_params_3C
        )

        x_a_image_3C = None
        if depth_map is not None:
            zB_idx = list(setup_3C.fit_names).index('zB')
            x_a_image_3C = np.broadcast_to(
                np.array(setup_3C.x_a), (*depth_map.shape, len(setup_3C.fit_names))
            ).copy()
            x_a_image_3C[..., zB_idx] = np.log(depth_map)

        Rrs_arr = rrs.transpose('y', 'x', 'band').values
        start = timeit.default_timer()
        results_3C = oe_engine_optx.invert_image_optx(
            Rrs_arr, setup_3C, noise=NOISE_STEP1, max_steps=MAX_STEPS, tile_size=TILE_SIZE,
            x_a_image=x_a_image_3C,
        )
        print(f'  Step 1 inversion done in {timeit.default_timer() - start:.1f}s')

        glint_params_ds = dask_oe_engine.to_dataset(
            results_3C, coords={'y': rrs.y, 'x': rrs.x},
        ).rio.write_crs(scene_crs)
        store.write_data(
            glint_params_ds,
            f"{OUTPUT_PREFIX}{scene_id}-glint_params.zarr", replace=True,
        )

        glint_arr = forward_glint_2D(glint_params_ds, params_3C, wavelengths, pre_3C)
        glint_da  = xr.DataArray(
            glint_arr, coords=rrs.coords, dims=rrs.dims,
        ).rio.write_crs(scene_crs)
        store.write_data(
            glint_da.to_dataset(name='glint'),
            f"{OUTPUT_PREFIX}{scene_id}-glint.zarr", replace=True,
        )

        Rrs_corr = (rrs - glint_da).rio.write_crs(scene_crs)
        store.write_data(Rrs_corr.to_dataset(name='Rrs'), rrs_path, replace=True)
        print(f'  Step 1 done. Saved: {rrs_path}')

    # =========================================================================
    # STEP 2 — Water quality retrieval (oe_engine_optx: lax.while_loop, early exit)
    # =========================================================================
    if need_step2:
        print('  Step 2: water quality retrieval ...')

        _rrs_ds    = store.open_data(rrs_path)
        y_coords   = _rrs_ds['Rrs'].y
        x_coords   = _rrs_ds['Rrs'].x
        Rrs_corr_arr = _rrs_ds['Rrs'].transpose('y', 'x', 'band').values

        pre_wq   = albert_mobley_jax.precompute(wavelengths)
        f_vec_wq = albert_mobley_jax.make_forward_vec(list(params_wq.keys()), pre_wq)
        setup_wq = oe_engine.build_inversion(
            params_wq, f_vec_wq, sigma_a_wq, log_params=log_params_wq
        )

        x_a_image_wq = None
        if depth_map is not None:
            zB_idx = list(setup_wq.fit_names).index('zB')
            x_a_image_wq = np.broadcast_to(
                np.array(setup_wq.x_a), (*depth_map.shape, len(setup_wq.fit_names))
            ).copy()
            x_a_image_wq[..., zB_idx] = np.log(depth_map)

        start = timeit.default_timer()
        results_wq = oe_engine_optx.invert_image_optx(
            Rrs_corr_arr, setup_wq, noise=NOISE_STEP2, max_steps=MAX_STEPS, tile_size=TILE_SIZE,
            x_a_image=x_a_image_wq,
        )
        print(f'  Step 2 inversion done in {timeit.default_timer() - start:.1f}s')

        wq_ds = dask_oe_engine.to_dataset(
            results_wq, coords={'y': y_coords, 'x': x_coords},
        ).rio.write_crs(scene_crs)
        store.write_data(wq_ds, wq_path, replace=True)
        print(f'  Step 2 done. Saved: {wq_path}')
```

```python
_scene = SCENE_IDS[0]

# --- Step 1 chi² (glint correction) ---
_gp   = store.open_data(f"{OUTPUT_PREFIX}{_scene}-glint_params.zarr")
_chi2_step1 = _gp['chi2'].values
_med1 = float(np.nanmedian(_chi2_step1[np.isfinite(_chi2_step1)]))
_noise_step1 = NOISE_STEP1 * np.sqrt(_med1)

# --- Step 2 chi² (water quality) ---
_wq   = store.open_data(f"{OUTPUT_PREFIX}{_scene}-wq_params.zarr")
_chi2_step2 = _wq['chi2'].values
_med2 = float(np.nanmedian(_chi2_step2[np.isfinite(_chi2_step2)]))
_noise_step2 = NOISE_STEP2 * np.sqrt(_med2)

print(f"Current noise:  NOISE_STEP1 = {NOISE_STEP1:.5f} sr⁻¹   NOISE_STEP2 = {NOISE_STEP2:.5f} sr⁻¹")
print()
print(f"Step 1 (3C / glint):  median chi² = {_med1:.2f}  →  calibrated NOISE_STEP1 = {_noise_step1:.5f} sr⁻¹")
print(f"Step 2 (WQ):          median chi² = {_med2:.2f}  →  calibrated NOISE_STEP2 = {_noise_step2:.5f} sr⁻¹")
print()
if _med2 > _med1:
    print("Step 2 chi² > Step 1 — expected: glint correction adds residual uncertainty.")
else:
    print("Step 2 chi² ≤ Step 1 — glint correction may have over-smoothed; check Rrs maps.")
print()
print("→ Update NOISE_STEP1 / NOISE_STEP2 in the params cell and delete the cached zarrs to re-run.")
```
